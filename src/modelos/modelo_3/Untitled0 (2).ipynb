```python
# -*- coding: utf-8 -*-
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, f1_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
import warnings
import re
import pickle
import time
import traceback

# Import para Correlação de Distância (instale com: pip install dcor)
try:
    import dcor
except ImportError:
    print("Aviso: A biblioteca 'dcor' não está instalada. A análise de correlação de distância será pulada.")
    print("Para instalá-la, execute: pip install dcor")
    dcor = None

# Imports para Ray Tune
import ray
from ray import tune
from ray.tune.schedulers import ASHAScheduler
from ray.tune.search.hyperopt import HyperOptSearch # Requer: pip install hyperopt

# Imports para Keras (TensorFlow)
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, Flatten, Dense, Dropout, concatenate, BatchNormalization, Activation
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras import regularizers # Para L2
from ray.train.tensorflow.keras import ReportCheckpointCallback

warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=FutureWarning)

os.makedirs('visualizacoes_classificacao_salario_v8_nn_final', exist_ok=True)

# --- Funções de limpeza de dados (sem alterações) ---
def clean_col_name(col_name):
    original_input = col_name
    if isinstance(col_name, tuple): col_name = "_".join(str(item).strip() for item in col_name)
    elif not isinstance(col_name, str): col_name = str(col_name)
    col_name = re.sub(r'[^\w\s-]', '', col_name).strip()
    col_name = re.sub(r'[-\s]+', '_', col_name); col_name = re.sub(r"_+", "_", col_name)
    col_name = col_name.strip("_");
    if not col_name: return f"col_limpa_vazia_{hash(original_input)}"
    if col_name and col_name[0].isdigit(): col_name = "_" + col_name
    return col_name

def extract_salary_lower_bound(salary_range_str):
    if pd.isna(salary_range_str): return np.nan
    s = str(salary_range_str).lower().replace('r$', '').replace('.', '').replace('/mês', '').strip()
    match_de_a = re.search(r'de\s*(\d+)\s*a\s*(\d+)', s)
    if match_de_a: return float(match_de_a.group(1))
    match_acima_de = re.search(r'acima de\s*(\d+)', s)
    if match_acima_de: return float(match_acima_de.group(1))
    match_menos_de = re.search(r'menos de\s*(\d+)', s)
    if match_menos_de: return 0
    match_so_numeros = re.findall(r'\d+', s)
    if match_so_numeros: return float(match_so_numeros[0])
    return np.nan

def map_uf_to_region(uf_series: pd.Series) -> pd.Series:
    mapa_regioes = {
        'AC': 'Norte', 'AL': 'Nordeste', 'AP': 'Norte', 'AM': 'Norte', 'BA': 'Nordeste',
        'CE': 'Nordeste', 'DF': 'Centro-Oeste', 'ES': 'Sudeste', 'GO': 'Centro-Oeste',
        'MA': 'Nordeste', 'MT': 'Centro-Oeste', 'MS': 'Centro-Oeste', 'MG': 'Sudeste',
        'PA': 'Norte', 'PB': 'Nordeste', 'PR': 'Sul', 'PE': 'Nordeste', 'PI': 'Nordeste',
        'RJ': 'Sudeste', 'RN': 'Nordeste', 'RS': 'Sul', 'RO': 'Norte', 'RR': 'Norte',
        'SC': 'Sul', 'SP': 'Sudeste', 'SE': 'Nordeste', 'TO': 'Norte'
    }
    uf_series_normalized = uf_series.astype(str).str.upper().str.strip()
    siglas_uf = list(mapa_regioes.keys())
    def extract_sigla(val):
        if val in siglas_uf: return val
        for sigla in siglas_uf:
            if f"({sigla})" in val or f" {sigla} " in val or val.endswith(f" {sigla}"): return sigla
        if "DISTRITO FEDERAL" in val: return "DF"
        if "SAO PAULO" in val: return "SP"
        if "RIO DE JANEIRO" in val: return "RJ"
        return val
    uf_series_normalized = uf_series_normalized.apply(extract_sigla)
    mapped_series = uf_series_normalized.map(mapa_regioes)
    return mapped_series.fillna('Desconhecida')

def clean_experience_to_numeric(exp_val):
    if pd.isna(exp_val): return np.nan
    s = str(exp_val).lower().strip()
    if 'menos de 1 ano' in s or 'menos de um ano' in s or '< 1 ano' in s: return 0.5
    if 'não tenho experiência' in s or 'sem experiência' in s: return 0
    numbers = re.findall(r'\d+\.?\d*', s)
    if numbers: return float(numbers[0])
    return np.nan
# --- [FIM DAS Funções de limpeza de dados] ---

# --- Função de criação do modelo Keras (CORRIGIDA) ---
def create_keras_model_v2_fixed(embedding_infos_list, num_numerical_features, config_nn):
    inputs = []; embedded_layers = []
    l2_strength_emb = config_nn.get("l2_strength_embedding", 0.0001)

    for emb_info in embedding_infos_list:
        input_cat = Input(shape=(emb_info['input_length'],), name=f"input_{emb_info['name']}")
        inputs.append(input_cat)
        embedding_output_dim = config_nn.get(f"emb_dim_{emb_info['name']}", emb_info['output_dim'])
        embedding = Embedding(input_dim=emb_info['input_dim'],
                              output_dim=embedding_output_dim,
                              name=f"embedding_{emb_info['name']}",
                              embeddings_regularizer=regularizers.l2(l2_strength_emb))(input_cat)
        flattened_embedding = Flatten(name=f"flatten_{emb_info['name']}")(embedding)
        embedded_layers.append(flattened_embedding)

    if num_numerical_features > 0:
        input_num = Input(shape=(num_numerical_features,), name="input_numerical")
        inputs.append(input_num)
        to_concatenate = embedded_layers + [input_num]
    else: to_concatenate = embedded_layers

    if len(to_concatenate) > 1: merged_features = concatenate(to_concatenate)
    elif to_concatenate: merged_features = to_concatenate[0]
    else: raise ValueError("O modelo precisa de pelo menos uma feature de entrada.")

    l2_strength_dense = config_nn.get("l2_strength_dense", 0.001)
    x = merged_features
    num_layers = config_nn.get("num_hidden_layers", 1)

    # CORREÇÃO: Loop para criar o número correto de camadas
    for i in range(1, num_layers + 1):
        dense_units = config_nn.get(f"dense_units_{i}", 64) # Pega dense_units_1, dense_units_2, etc.
        dropout_rate = config_nn.get(f"dropout_{i}", 0.3) # Pega dropout_1, dropout_2, etc.
        x = Dense(dense_units, kernel_regularizer=regularizers.l2(l2_strength_dense))(x)
        x = BatchNormalization()(x)
        x = Activation("relu")(x)
        x = Dropout(dropout_rate)(x)

    output = Dense(1, activation="sigmoid", name="output")(x)
    model = Model(inputs=inputs, outputs=output)

    optimizer_type = config_nn.get("optimizer", "adam")
    learning_rate = config_nn.get("learning_rate_nn", 0.001)
    if optimizer_type == "adamw":
        optimizer = tf.keras.optimizers.AdamW(learning_rate=learning_rate, weight_decay=config_nn.get("weight_decay", 0.004))
    elif optimizer_type == "nadam":
        optimizer = tf.keras.optimizers.Nadam(learning_rate=learning_rate)
    else: # Default to Adam
        optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate)

    model.compile(optimizer=optimizer, loss="binary_crossentropy", metrics=["accuracy"])
    return model

# --- Função Objetivo para Ray Tune ---
def objective_ray_tune_keras_v2(config_keras, X_train_list, y_train_arr, X_val_list, y_val_arr,
                                embedding_infos_obj, num_numerical_features_obj):
    tf.keras.backend.clear_session()
    model = create_keras_model_v2_fixed(embedding_infos_obj, num_numerical_features_obj, config_keras)

    early_stopping = EarlyStopping(monitor='val_accuracy', patience=config_keras.get("early_stopping_patience", 10),
                                   restore_best_weights=True, mode='max')
    report_checkpoint_callback = ReportCheckpointCallback(metrics={"val_accuracy_tune": "val_accuracy"})
    reduce_lr_callback = tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=config_keras.get("reduce_lr_factor", 0.2),
                                                              patience=config_keras.get("reduce_lr_patience", 3), min_lr=1e-6)

    model.fit(
        X_train_list, y_train_arr,
        validation_data=(X_val_list, y_val_arr),
        epochs=config_keras.get("epochs", 50),
        batch_size=config_keras.get("batch_size", 32),
        callbacks=[early_stopping, report_checkpoint_callback, reduce_lr_callback],
        verbose=0
    )

# --- Função para Importância por Permutação ---
def calculate_permutation_importance(model, X_test_list, y_test, baseline_metric, feature_names_cat, feature_names_num):
    print("\n--- Calculando Importância de Features por Permutação ---")
    importances = {}
    full_feature_list = feature_names_cat + feature_names_num

    # Índice do array numérico na lista de inputs
    num_array_idx = len(feature_names_cat) if feature_names_num else -1

    for i, feature_name in enumerate(full_feature_list):
        X_test_permuted = [x.copy() for x in X_test_list]

        if feature_name in feature_names_cat:
            # Feature categórica: o índice 'i' corresponde ao seu array na lista
            permuted_col = np.random.permutation(X_test_permuted[i])
            X_test_permuted[i] = permuted_col

        elif feature_name in feature_names_num and num_array_idx != -1:
            # Feature numérica: encontrar sua coluna dentro do array numérico
            num_feature_idx = feature_names_num.index(feature_name)
            permuted_col = np.random.permutation(X_test_permuted[num_array_idx][:, num_feature_idx])
            X_test_permuted[num_array_idx][:, num_feature_idx] = permuted_col
        else:
            continue

        y_pred_permuted = (model.predict(X_test_permuted) > 0.5).astype(int)
        permuted_metric = accuracy_score(y_test, y_pred_permuted)
        importance = baseline_metric - permuted_metric
        importances[feature_name] = importance
        print(f"Importância de '{feature_name}': {importance:.4f}")

    return importances

# --- Função Principal de Treinamento ---
def train_classification_model_salary_range_v8_final(file_path, n_ray_tune_samples_nn=50, ray_tune_timeout_nn=3600):
    if not ray.is_initialized():
        num_cpus_to_use = max(1, os.cpu_count() - 1 if os.cpu_count() else 1)
        print(f"Inicializando Ray com {num_cpus_to_use} CPUs...")
        ray.init(ignore_reinit_error=True, num_cpus=num_cpus_to_use)

    try:
        df_main = pd.read_excel(file_path)
        df_main.columns = [clean_col_name(col) for col in df_main.columns]

        # Mapeamento de colunas (simplificado para clareza)
        col_mapping = {
            "target_original_salary_range": next((c for c in df_main.columns if "P2_h" in c or "faixa_salarial" in c.lower()), None),
            "tempo_experiencia_P2i": next((c for c in df_main.columns if "P2_i" in c), None),
            "nivel_senioridade_P2g": next((c for c in df_main.columns if "P2_g" in c), None),
            "uf_mora_P1i1": next((c for c in df_main.columns if "P1_i_1" in c), None),
            "cargo_atual_P2f": next((c for c in df_main.columns if "P2_f" in c), None),
            "faixa_etaria": next((c for c in df_main.columns if "P1_a_1" in c or "faixa_etaria" in c.lower()), None),
            "genero": next((c for c in df_main.columns if "P1_b" in c or "genero" in c.lower()), None),
            "nivel_ensino": next((c for c in df_main.columns if "P1_l" in c or "nivel_ensino" in c.lower()), None),
        }
        col_mapping = {k: v for k, v in col_mapping.items() if v} # Remover nulos

        # Pré-processamento e criação da variável alvo
        df_main[col_mapping["tempo_experiencia_P2i"]] = df_main[col_mapping["tempo_experiencia_P2i"]].apply(clean_experience_to_numeric)
        df_main_processed = df_main.dropna(subset=[col_mapping["target_original_salary_range"]]).copy()
        df_main_processed['salary_numeric_lower_bound'] = df_main_processed[col_mapping["target_original_salary_range"]].apply(extract_salary_lower_bound)
        df_main_processed.dropna(subset=['salary_numeric_lower_bound'], inplace=True)

        target_col_agrupada_name = "faixa_salarial_agrupada"
        point_of_cut_fixed = 7500.0
        df_main_processed[target_col_agrupada_name] = np.where(df_main_processed['salary_numeric_lower_bound'] >= point_of_cut_fixed, 'Salário Alto', 'Salário Baixo')
        print(f"Contagens da variável alvo '{target_col_agrupada_name}':\n{df_main_processed[target_col_agrupada_name].value_counts(dropna=False)}")

        # Seleção e preparação das features
        features_para_nn_originais = [
            "faixa_etaria", "genero", "nivel_ensino", "tempo_experiencia_P2i",
            "nivel_senioridade_P2g", "uf_mora_P1i1", "cargo_atual_P2f"
        ]
        feature_cols_nn = [col_mapping[col] for col in features_para_nn_originais if col in col_mapping]
        X_initial_nn = df_main_processed[feature_cols_nn].copy()

        if col_mapping.get("uf_mora_P1i1") in X_initial_nn.columns:
            X_initial_nn['Regiao_Mapeada'] = map_uf_to_region(X_initial_nn[col_mapping["uf_mora_P1i1"]])
            X_initial_nn.drop(columns=[col_mapping["uf_mora_P1i1"]], inplace=True)

        le_target = LabelEncoder()
        y_full_nn = pd.Series(le_target.fit_transform(df_main_processed[target_col_agrupada_name]), index=X_initial_nn.index)

        # Tratamento de Outliers (Apenas em numéricas)
        numerical_cols = X_initial_nn.select_dtypes(include=np.number).columns
        if not numerical_cols.empty:
            Q1 = X_initial_nn[numerical_cols].quantile(0.25)
            Q3 = X_initial_nn[numerical_cols].quantile(0.75)
            IQR = Q3 - Q1
            outliers = ((X_initial_nn[numerical_cols] < (Q1 - 1.5 * IQR)) | (X_initial_nn[numerical_cols] > (Q3 + 1.5 * IQR))).any(axis=1)
            print(f"Removendo {outliers.sum()} outliers das features numéricas.")
            X_initial_nn = X_initial_nn[~outliers]; y_full_nn = y_full_nn[~outliers]

        # --- EDA: Análise de Correlação ---
        print("\n--- Gerando Gráficos de Correlação ---")
        df_corr_analysis = X_initial_nn.copy()
        # Para análise, transformar todas as categóricas em numéricas temporariamente
        temp_encoders = {}
        for col in df_corr_analysis.select_dtypes(include='object').columns:
            temp_encoders[col] = LabelEncoder()
            df_corr_analysis[col] = temp_encoders[col].fit_transform(df_corr_analysis[col].astype(str))

        # 1. Correlação de Pearson
        plt.figure(figsize=(12, 9))
        sns.heatmap(df_corr_analysis.corr(), annot=True, cmap='viridis', fmt='.2f')
        plt.title('Heatmap de Correlação de Pearson', fontsize=16)
        plt.show()

        # 2. Correlação de Distância (se 'dcor' estiver instalado)
        if dcor:
            try:
                # CORREÇÃO: Usar dcor.pairwise.distance_correlation_pairwise
                dcor_features = df_corr_analysis.columns.tolist()
                print("Calculando a matriz de correlação de distância (pode levar um momento)...")
                # A função correta aceita o DataFrame diretamente
                dcor_matrix = dcor.pairwise.distance_correlation_pairwise(df_corr_analysis[dcor_features])

                plt.figure(figsize=(12, 9))
                sns.heatmap(dcor_matrix, annot=True, cmap='rocket', fmt='.2f',
                            xticklabels=dcor_features, yticklabels=dcor_features)
                plt.title('Heatmap de Correlação de Distância (dcor)', fontsize=16)
                plt.show()
            except Exception as e:
                print(f"Ocorreu um erro ao gerar o gráfico de correlação de distância: {e}")

        # --- Preparação dos dados para Rede Neural ---
        X_train_nn_full, X_test_nn_full, y_train_nn_full, y_test_nn_full = train_test_split(
            X_initial_nn, y_full_nn, test_size=0.25, random_state=42, stratify=y_full_nn)

        numerical_features_nn = X_train_nn_full.select_dtypes(include=np.number).columns.tolist()
        categorical_features_nn = X_train_nn_full.select_dtypes(exclude=np.number).columns.tolist()

        scaler_nn_final = StandardScaler()
        X_train_numerical_scaled = scaler_nn_final.fit_transform(X_train_nn_full[numerical_features_nn]) if numerical_features_nn else np.array([]).reshape(len(X_train_nn_full), 0)
        X_test_numerical_scaled = scaler_nn_final.transform(X_test_nn_full[numerical_features_nn]) if numerical_features_nn else np.array([]).reshape(len(X_test_nn_full), 0)

        label_encoders_nn = {}; embedding_infos_final = []; unk_category_map = {}
        X_train_categorical_list = []; X_test_categorical_list = []

        for col in categorical_features_nn:
            le = LabelEncoder()
            train_series = X_train_nn_full[col].astype(str).fillna("Missing")
            test_series = X_test_nn_full[col].astype(str).fillna("Missing")
            le.fit(train_series)

            # Adicionar categoria 'Unknown' para valores não vistos no treino
            classes = list(le.classes_) + ['Unknown']
            le.classes_ = np.array(classes)
            unk_idx = len(le.classes_) - 1
            unk_category_map[col] = unk_idx

            train_encoded = train_series.map(lambda s: le.transform([s])[0] if s in le.classes_ else unk_idx)
            test_encoded = test_series.map(lambda s: le.transform([s])[0] if s in le.classes_ else unk_idx)

            X_train_categorical_list.append(train_encoded.values.reshape(-1, 1))
            X_test_categorical_list.append(test_encoded.values.reshape(-1, 1))

            cardinality = len(le.classes_)
            embedding_size = min(50, int(np.ceil(cardinality**0.25) * 4))
            embedding_infos_final.append({'name': col, 'input_dim': cardinality, 'output_dim': embedding_size, 'input_length': 1})
            label_encoders_nn[col] = le

        X_train_nn_inputs_final_list = X_train_categorical_list + ([X_train_numerical_scaled] if numerical_features_nn else [])
        X_test_nn_inputs_final_list = X_test_categorical_list + ([X_test_numerical_scaled] if numerical_features_nn else [])
        y_train_nn_final_arr = y_train_nn_full.values
        y_test_nn_final_arr = y_test_nn_full.values

        # Split para HPO
        train_hpo_idx, val_hpo_idx = train_test_split(np.arange(len(y_train_nn_final_arr)), test_size=0.2, random_state=42, stratify=y_train_nn_final_arr)
        X_train_hpo_list = [data[train_hpo_idx] for data in X_train_nn_inputs_final_list]
        y_train_hpo_arr = y_train_nn_final_arr[train_hpo_idx]
        X_val_hpo_list = [data[val_hpo_idx] for data in X_train_nn_inputs_final_list]
        y_val_hpo_arr = y_train_nn_final_arr[val_hpo_idx]

        # --- Espaço de Busca Ray Tune (ATUALIZADO) ---
        search_space_nn_ray = {
            "num_hidden_layers": tune.choice([1, 2, 3]),
            "dense_units_1": tune.choice([64, 128, 256, 512]),
            "dropout_1": tune.quniform(0.1, 0.5, 0.05),
            "dense_units_2": tune.choice([32, 64, 128, 256]),
            "dropout_2": tune.quniform(0.1, 0.5, 0.05),
            "dense_units_3": tune.choice([16, 32, 64, 128]), # Para a 3ª camada
            "dropout_3": tune.quniform(0.1, 0.4, 0.05),   # Para a 3ª camada
            "learning_rate_nn": tune.loguniform(1e-4, 5e-3),
            "batch_size": tune.choice([32, 64, 128]),
            "epochs": 100, # Fixo, ES controla
            "early_stopping_patience": tune.choice([10, 15]),
            "l2_strength_embedding": tune.loguniform(1e-5, 1e-2),
            "l2_strength_dense": tune.loguniform(1e-5, 1e-2),
            "optimizer": tune.choice(["adam", "nadam", "adamw"]),
            "weight_decay": tune.loguniform(1e-5, 1e-2),
            "reduce_lr_patience": tune.choice([3, 5]),
            "reduce_lr_factor": tune.choice([0.2, 0.5])
        }
        for emb_info in embedding_infos_final:
             search_space_nn_ray[f"emb_dim_{emb_info['name']}"] = tune.choice([max(4, emb_info['output_dim'] // 2), emb_info['output_dim']])

        asha_scheduler = ASHAScheduler(metric="val_accuracy_tune", mode='max', max_t=100, grace_period=15, reduction_factor=2)
        search_algo_nn = HyperOptSearch(metric="val_accuracy_tune", mode="max")

        print(f"\nIniciando otimização da RNA com Ray Tune ({n_ray_tune_samples_nn} amostras)...")
        analysis_nn = tune.run(
            tune.with_parameters(
                objective_ray_tune_keras_v2,
                X_train_list=X_train_hpo_list, y_train_arr=y_train_hpo_arr,
                X_val_list=X_val_hpo_list, y_val_arr=y_val_hpo_arr,
                embedding_infos_obj=embedding_infos_final,
                num_numerical_features_obj=X_train_numerical_scaled.shape[1]
            ),
            config=search_space_nn_ray, num_samples=n_ray_tune_samples_nn,
            scheduler=asha_scheduler, search_alg=search_algo_nn,
            resources_per_trial={"cpu": 1, "gpu": 0},
            storage_path=os.path.abspath("./ray_results_salary_v8_final"),
            name=f"Keras_Salary_Clf_RayTune_v8",
            verbose=1, raise_on_failed_trial=False,
            time_budget_s=ray_tune_timeout_nn
        )

        best_trial_nn_ray = analysis_nn.get_best_trial(metric="val_accuracy_tune", mode="max", scope="all")
        if not best_trial_nn_ray:
            print("ERRO CRÍTICO: Nenhum trial do Ray Tune foi bem-sucedido.")
            return None

        best_nn_params_hpo = best_trial_nn_ray.config
        print(f"\nMelhores hiperparâmetros (RNA Ray Tune): {best_nn_params_hpo}")

        # Treinamento do Modelo Final
        print("\nTreinando o modelo RNA final com os melhores hiperparâmetros...")
        final_nn_model = create_keras_model_v2_fixed(embedding_infos_final, X_train_numerical_scaled.shape[1], best_nn_params_hpo)

        final_early_stopping = EarlyStopping(monitor='val_accuracy', patience=20, restore_best_weights=True, mode='max', verbose=1)
        final_reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=7, min_lr=1e-7, verbose=1)

        final_nn_model.fit(
            X_train_nn_inputs_final_list, y_train_nn_final_arr,
            validation_data=(X_test_nn_inputs_final_list, y_test_nn_final_arr), # Validando no teste para ES
            epochs=200, batch_size=best_nn_params_hpo.get("batch_size", 64),
            callbacks=[final_early_stopping, final_reduce_lr], verbose=1
        )

        # Avaliação Final e Salvamento
        print("\nAvaliando o modelo RNA final no conjunto de teste...")
        y_pred_proba_test_nn = final_nn_model.predict(X_test_nn_inputs_final_list)
        y_pred_test_nn = (y_pred_proba_test_nn > 0.5).astype(int)

        accuracy_test_nn = accuracy_score(y_test_nn_final_arr, y_pred_test_nn)
        f1_test_nn = f1_score(y_test_nn_final_arr, y_pred_test_nn, average='weighted')
        roc_auc_test_nn = roc_auc_score(y_test_nn_final_arr, y_pred_proba_test_nn)

        print(f"\n--- Resultados da Avaliação da REDE NEURAL no CONJUNTO DE TESTE ---")
        print(f"Acurácia no Teste: {accuracy_test_nn:.4f}")
        print(f"F1-Score (Ponderado) no Teste: {f1_test_nn:.4f}")
        print(f"ROC AUC no Teste: {roc_auc_test_nn:.4f}")

        plt.figure(figsize=(8, 6))
        sns.heatmap(confusion_matrix(y_test_nn_final_arr, y_pred_test_nn), annot=True, fmt="d", cmap="Blues", xticklabels=le_target.classes_, yticklabels=le_target.classes_)
        plt.title('Matriz de Confusão (Teste - RNA Final)')
        plt.savefig('visualizacoes_classificacao_salario_v8_nn_final/matriz_confusao_nn_teste.png')
        plt.show()

        # --- Importância de Features por Permutação ---
        importances = calculate_permutation_importance(
            final_nn_model, X_test_nn_inputs_final_list, y_test_nn_final_arr,
            accuracy_test_nn, categorical_features_nn, numerical_features_nn)

        df_importance = pd.DataFrame.from_dict(importances, orient='index', columns=['importance'])
        df_importance.sort_values(by='importance', ascending=False, inplace=True)

        plt.figure(figsize=(10, 8))
        sns.barplot(x='importance', y=df_importance.index, data=df_importance, palette='cubehelix')
        plt.title('Importância de Features por Permutação (no Conjunto de Teste)', fontsize=16)
        plt.xlabel('Queda na Acurácia')
        plt.ylabel('Feature')
        plt.grid(axis='x', linestyle='--', alpha=0.7)
        plt.show()

        # Salvar modelo e pré-processadores
        model_filename = 'modelo_keras_salario_v8.keras'
        preprocessors_filename = 'preprocessors_nn_salario_v8.pkl'
        final_nn_model.save(model_filename)
        with open(preprocessors_filename, 'wb') as f:
            pickle.dump({
                'label_encoders_nn': label_encoders_nn, 'scaler_nn_final': scaler_nn_final,
                'embedding_infos_final': embedding_infos_final,
                'feature_column_names_in_order': X_initial_nn.columns.tolist(),
                'numerical_features_nn': numerical_features_nn, 'categorical_features_nn': categorical_features_nn,
                'label_encoder_target': le_target, 'unk_category_map': unk_category_map
            }, f)
        print(f"Modelo salvo em {model_filename}")
        print(f"Pré-processadores salvos em {preprocessors_filename}")

        return {'nn_accuracy_test': accuracy_test_nn, 'nn_model_path': model_filename}
    except Exception as e:
        print(f"Ocorreu um erro inesperado: {e}")
        traceback.print_exc()
        return None
    finally:
        if ray.is_initialized():
            print("Desligando Ray...")
            ray.shutdown()

if __name__ == "__main__":
    file_path = "Main_database (2).xlsx" # Substitua pelo caminho do seu arquivo
    if os.path.exists(file_path):
        results = train_classification_model_salary_range_v8_final(
            file_path,
            n_ray_tune_samples_nn=75,
            ray_tune_timeout_nn=7200 # 2 horas
        )
        if results:
            print("\n--- Execução Concluída com Sucesso ---")
    else:
        print(f"ERRO: Arquivo '{file_path}' não encontrado.")
```
