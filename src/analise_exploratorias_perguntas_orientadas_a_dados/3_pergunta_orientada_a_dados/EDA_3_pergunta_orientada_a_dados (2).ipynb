```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re
import plotly.express as px
# from ydata_profiling import ProfileReport
import statsmodels.api as sm

# --- Funções de pré-processamento (Mantidas) ---
def clean_col_name(col_name):
    original_input = col_name
    if isinstance(col_name, tuple):
        col_name = "_".join(str(item).strip() for item in col_name)
    elif not isinstance(col_name, str):
        col_name = str(col_name)
    col_name = re.sub(r'[^\w\s-]', '', col_name).strip()
    col_name = re.sub(r'[-\s]+', '_', col_name)
    col_name = re.sub(r"_+", "_", col_name)
    col_name = col_name.strip("_")
    if not col_name: return f"col_limpa_vazia_{hash(original_input)}"
    if col_name and col_name[0].isdigit(): col_name = "_" + col_name
    return col_name

def extract_salary_lower_bound(salary_range_str):
    if pd.isna(salary_range_str): return np.nan
    s = str(salary_range_str).lower().replace('r$', '').replace('.', '').replace('/mês', '').strip()
    match_de_a = re.search(r'de\s*(\d+)\s*a\s*(\d+)', s)
    if match_de_a: return float(match_de_a.group(1))
    match_acima_de = re.search(r'acima de\s*(\d+)', s)
    if match_acima_de: return float(match_acima_de.group(1))
    match_menos_de = re.search(r'menos de\s*(\d+)', s)
    if match_menos_de: return 0
    match_so_numeros = re.findall(r'\d+', s)
    if match_so_numeros: return float(match_so_numeros[0])
    return np.nan

def map_uf_to_region(uf_series: pd.Series) -> pd.Series:
    mapa_regioes = {
        'AC': 'Norte', 'AL': 'Nordeste', 'AP': 'Norte', 'AM': 'Norte', 'BA': 'Nordeste',
        'CE': 'Nordeste', 'DF': 'Centro-Oeste', 'ES': 'Sudeste', 'GO': 'Centro-Oeste',
        'MA': 'Nordeste', 'MT': 'Centro-Oeste', 'MS': 'Centro-Oeste', 'MG': 'Sudeste',
        'PA': 'Norte', 'PB': 'Nordeste', 'PR': 'Sul', 'PE': 'Nordeste', 'PI': 'Nordeste',
        'RJ': 'Sudeste', 'RN': 'Nordeste', 'RS': 'Sul', 'RO': 'Norte', 'RR': 'Norte',
        'SC': 'Sul', 'SP': 'Sudeste', 'SE': 'Nordeste', 'TO': 'Norte'
    }
    uf_series_normalized = uf_series.astype(str).str.upper().str.strip()
    siglas_uf = list(mapa_regioes.keys())
    def extract_sigla(val):
        if val in siglas_uf: return val
        for sigla in siglas_uf:
            if f"({sigla})" in val or f" {sigla} " in val or val.endswith(f" {sigla}"): return sigla
        if "DISTRITO FEDERAL" in val: return "DF"
        if "SAO PAULO" in val: return "SP"
        if "RIO DE JANEIRO" in val: return "RJ"
        if "MINAS GERAIS" in val: return "MG"
        if "ESPIRITO SANTO" in val: return "ES"
        if "RIO GRANDE DO SUL" in val: return "RS"
        if "SANTA CATARINA" in val: return "SC"
        if "PARANA" in val: return "PR"
        return val
    uf_series_normalized = uf_series_normalized.apply(extract_sigla)
    mapped_series = uf_series_normalized.map(mapa_regioes)
    return mapped_series.fillna('Desconhecida')

def clean_experience_to_numeric(exp_val):
    if pd.isna(exp_val):
        return np.nan
    s = str(exp_val).lower().strip()
    if 'menos de 1 ano' in s or 'menos de um ano' in s or '< 1 ano' in s:
        return 0.5
    if 'não tenho experiência' in s or 'sem experiência' in s:
        return 0
    numbers = re.findall(r'\d+\.?\d*', s)
    if numbers:
        return float(numbers[0])
    return np.nan

# --- Configurações da EDA ---
eda_output_dir_script = 'visualizacoes_eda_script_univariada_final'
os.makedirs(eda_output_dir_script, exist_ok=True)
sns.set_style("whitegrid")

# --- 1. Carregar Dados ---
print("--- 1. Carregando Dados ---")
file_path = "Main_database (2).xlsx"
if not os.path.exists(file_path):
    print(f"ERRO: Arquivo de dados '{file_path}' não encontrado.")
    exit()
df_original = pd.read_excel(file_path)
print(f"Base de dados original carregada: {df_original.shape[0]} linhas, {df_original.shape[1]} colunas.")

# --- 2. Limpeza Inicial de Nomes de Colunas ---
print("\n--- 2. Limpando Nomes de Colunas ---")
df_cleaned_names = df_original.copy()
df_cleaned_names.columns = [clean_col_name(col) for col in df_original.columns]
print("Nomes de colunas limpos.")

# --- 3. Seleção dos Atributos para EDA ---
print("\n--- 3. Selecionando Atributos para EDA ---")
col_identifiers = {
    'P1_a_1': 'P1_a_1', 'P1_b': 'P1_b', 'P1_l': 'P1_l',
    'P2_h': 'P2_h', 'P2_i': 'P2_i', 'P1_i_1': 'P1_i_1_uf_onde_mora',
    'P2_f': 'P2_f_Cargo_Atual', 'P2_g': 'P2_g_Nivel'
}
df_eda = pd.DataFrame()
for original_key, pattern in col_identifiers.items():
    found_col_name = next((cn for cn in df_cleaned_names.columns if pattern.lower() in cn.lower()), None)
    if found_col_name:
        df_eda[original_key] = df_cleaned_names[found_col_name]
        print(f"Coluna '{original_key}' (mapeada de '{found_col_name}') selecionada.")
    else:
        print(f"Aviso: Padrão '{pattern}' para '{original_key}' não encontrado.")
if df_eda.empty or 'P2_h' not in df_eda.columns:
    print("ERRO: Colunas essenciais para EDA não encontradas ou 'P2_h' ausente.")
    exit()
print(f"Shape do DataFrame de EDA inicial: {df_eda.shape}")

# --- 4. Limpeza e Transformação dos Atributos ---
print("\n--- 4. Limpando e Transformando Atributos ---")
if 'P2_h' in df_eda.columns:
    df_eda['salary_numeric_lower_bound'] = df_eda['P2_h'].apply(extract_salary_lower_bound)
    df_eda.dropna(subset=['salary_numeric_lower_bound'], inplace=True)
    if not df_eda.empty:
        min_salary_eda = df_eda['salary_numeric_lower_bound'].min()
        max_salary_eda = df_eda['salary_numeric_lower_bound'].max()
        point_of_cut_eda = 7500.0
        print(f"Usando ponto de corte para EDA: {point_of_cut_eda}")
        eda_salary_labels = ["Salário Baixo", "Salário Alto"]
        if min_salary_eda == max_salary_eda:
            df_eda['faixa_salarial_eda_2cat'] = eda_salary_labels[0]
        else:
            bins_eda, labels_eda = ([min_salary_eda, max_salary_eda], [eda_salary_labels[1]]) if point_of_cut_eda <= min_salary_eda else \
                                 ([min_salary_eda, max_salary_eda], [eda_salary_labels[0]]) if point_of_cut_eda >= max_salary_eda else \
                                 ([min_salary_eda, point_of_cut_eda, max_salary_eda], eda_salary_labels)
            unique_bins_eda = sorted(list(set(bins_eda)))
            if len(unique_bins_eda) < 2: unique_bins_eda = [min_salary_eda, max_salary_eda]
            if len(unique_bins_eda) == 2 and len(labels_eda) == 2:
                 labels_eda = [labels_eda[0]] if point_of_cut_eda >= (min_salary_eda + max_salary_eda)/2 else [labels_eda[1]]
            df_eda['faixa_salarial_eda_2cat'] = pd.cut(
                df_eda['salary_numeric_lower_bound'], bins=unique_bins_eda,
                labels=labels_eda, include_lowest=True, duplicates='drop'
            )
        df_eda.dropna(subset=['faixa_salarial_eda_2cat'], inplace=True)
        print(f"'faixa_salarial_eda_2cat' criada. Contas:\n{df_eda['faixa_salarial_eda_2cat'].value_counts(dropna=False)}")
    else: print("DataFrame vazio após processar 'salary_numeric_lower_bound'.")
else: print("ERRO: Coluna 'P2_h' não encontrada."); exit()

if 'P2_i' in df_eda.columns:
    df_eda['experiencia_anos'] = df_eda['P2_i'].apply(clean_experience_to_numeric)
    median_exp = df_eda['experiencia_anos'].median()
    df_eda['experiencia_anos'].fillna(median_exp, inplace=True)
    print(f"'experiencia_anos' criada. Nulos preenchidos com mediana ({median_exp:.1f}).")
if 'P1_i_1' in df_eda.columns:
    df_eda['Regiao_Mapeada'] = map_uf_to_region(df_eda['P1_i_1'])
categorical_cols_original_keys = ['P1_a_1', 'P1_b', 'P1_l', 'P2_f', 'P2_g']
for key in categorical_cols_original_keys:
    if key in df_eda.columns:
        df_eda[key] = df_eda[key].astype(str).fillna("Não Informado").str.strip()
        if df_eda[key].nunique() > 20:
            top_categories = df_eda[key].value_counts().nlargest(19).index
            df_eda[key] = df_eda[key].apply(lambda x: x if x in top_categories else 'Outros')
print(f"Shape do DataFrame de EDA final: {df_eda.shape}")
if df_eda.empty or 'faixa_salarial_eda_2cat' not in df_eda.columns:
    print("ERRO: DataFrame de EDA vazio ou sem coluna alvo."); exit()

# --- 6. Análise Univariada Expandida (Cada gráfico em uma imagem separada e exibido) ---
print("\n--- 6. Análise Univariada Expandida ---")
print("Gerando e salvando gráficos individuais...")

# 6.1 Variáveis Numéricas
numeric_cols_for_eda = ['salary_numeric_lower_bound', 'experiencia_anos']
for col in numeric_cols_for_eda:
    if col in df_eda.columns and pd.api.types.is_numeric_dtype(df_eda[col]) and not df_eda[col].dropna().empty:
        print(f"\n  Analisando numericamente '{col}':")
        print(df_eda[col].describe())

        plt.figure(figsize=(10, 6))
        sns.histplot(df_eda[col].dropna(), kde=False, bins=30, color="lightgreen", label='Histograma (Contagem)')
        ax_hist = plt.gca(); ax_kde = ax_hist.twinx()
        sns.kdeplot(df_eda[col].dropna(), color='darkblue', lw=2, label='Linha KDE Estimada (Densidade)', ax=ax_kde)
        ax_hist.set_title(f'Histograma e KDE de {col}', fontsize=15); ax_hist.set_xlabel(col, fontsize=12)
        ax_hist.set_ylabel('Contagem (Histograma)', fontsize=12, color='green'); ax_kde.set_ylabel('Densidade (KDE)', fontsize=12, color='darkblue')
        ax_hist.tick_params(axis='y', labelcolor='green'); ax_kde.tick_params(axis='y', labelcolor='darkblue')
        lines_hist, labels_hist = ax_hist.get_legend_handles_labels(); lines_kde, labels_kde = ax_kde.get_legend_handles_labels()
        ax_kde.legend(lines_hist + lines_kde, labels_hist + labels_kde, loc='upper right'); plt.tight_layout()
        plt.savefig(os.path.join(eda_output_dir_script, f'univar_numeric_{col}_hist_kde.png')); plt.show(); plt.close()

        plt.figure(figsize=(7, 6)); sns.boxplot(y=df_eda[col].dropna(), color="mediumaquamarine")
        plt.title(f'Boxplot de {col}', fontsize=15); plt.ylabel(col, fontsize=12); plt.tight_layout()
        plt.savefig(os.path.join(eda_output_dir_script, f'univar_numeric_{col}_boxplot.png')); plt.show(); plt.close()

        plt.figure(figsize=(8, 5)); sns.ecdfplot(data=df_eda.dropna(subset=[col]), x=col, color="darkblue")
        plt.title(f'ECDF de {col}', fontsize=15); plt.xlabel(col, fontsize=12); plt.ylabel('Proporção Cumulativa', fontsize=12)
        plt.grid(True, linestyle='--', alpha=0.7); plt.tight_layout()
        plt.savefig(os.path.join(eda_output_dir_script, f'univar_numeric_{col}_ecdf.png')); plt.show(); plt.close()

        data_for_qq = df_eda[col].dropna()
        if not data_for_qq.empty:
            fig_qq = plt.figure(figsize=(8, 7)); ax_qq = fig_qq.add_subplot(111)
            sm.qqplot(data_for_qq, line='s', ax=ax_qq)
            points = ax_qq.get_lines()[0]; points.set_markerfacecolor('lightgreen'); points.set_markeredgecolor('darkgreen'); points.set_marker('o')
            if len(ax_qq.get_lines()) > 1:
                line_ref = ax_qq.get_lines()[1]; line_ref.set_color('darkblue'); line_ref.set_linewidth(2); line_ref.set_label('Linha de Referência (Normal Teórica)')
            ax_qq.set_title(f'QQ-Plot de {col} vs Distribuição Normal', fontsize=15)
            ax_qq.set_xlabel('Quantis Teóricos (Distribuição Normal Padrão)', fontsize=12); ax_qq.set_ylabel('Quantis da Amostra (Ordenados)', fontsize=12)
            ax_qq.legend(loc='best'); plt.grid(True, linestyle='--', alpha=0.7); plt.tight_layout()
            plt.savefig(os.path.join(eda_output_dir_script, f'univar_numeric_{col}_qqplot_custom.png')); plt.show(); plt.close(fig_qq)
            print(f"  Gráfico QQ-Plot customizado para '{col}' salvo.")
        else: print(f"  Não há dados para gerar QQ-Plot para '{col}' após dropna.")
        print(f"  Gráficos numéricos para '{col}' salvos.")
    else: print(f"  Coluna '{col}' não é numérica, está vazia ou não encontrada. Pulando análise numérica.")

# 6.2 Variáveis Categóricas
categorical_cols_to_plot_univar = ['P1_a_1', 'P1_b', 'P1_l', 'P2_f', 'P2_g', 'Regiao_Mapeada', 'faixa_salarial_eda_2cat']
cat_palettes_seaborn = ["pastel", "Set2", "husl", "viridis", "plasma", "cividis", "rocket_r"]
cat_palettes_plotly = [px.colors.qualitative.Pastel, px.colors.qualitative.Set2, px.colors.qualitative.Vivid, px.colors.qualitative.Light24]

for i, col in enumerate(categorical_cols_to_plot_univar):
    if col in df_eda.columns and df_eda[col].nunique() > 0:
        print(f"\n  Analisando categoricamente '{col}':")
        value_counts = df_eda[col].value_counts(dropna=False)
        percentage_counts = df_eda[col].value_counts(normalize=True, dropna=False).mul(100).round(1)
        print("  Contagens:\n", value_counts); print("\n  Porcentagens:\n", percentage_counts.astype(str) + '%')

        plt.figure(figsize=(10, max(6, value_counts.shape[0] * 0.5)))
        order = value_counts.index
        current_palette_sns = cat_palettes_seaborn[i % len(cat_palettes_seaborn)]
        sns.countplot(y=df_eda[col], order=order, palette=current_palette_sns)
        plt.title(f'Distribuição de {col}', fontsize=15); plt.xlabel('Contagem', fontsize=12); plt.ylabel(col, fontsize=12)
        plt.tight_layout(); plt.savefig(os.path.join(eda_output_dir_script, f'univar_cat_{col}_countplot.png')); plt.show(); plt.close()
        print(f"  Gráfico de barras para '{col}' (paleta: {current_palette_sns}) salvo.")

        # Gráfico de Pizza (Matplotlib) - COM LEGENDA INFERIOR
        if value_counts.shape[0] <= 8 and value_counts.shape[0] > 1:
            fig_pie, ax_pie = plt.subplots(figsize=(8, 9)) # Ajustado para legenda inferior
            pie_colors_sns = sns.color_palette(cat_palettes_seaborn[(i+1) % len(cat_palettes_seaborn)], n_colors=value_counts.shape[0])
            labels_for_legend = [f'{label} ({percentage_counts[label]}%)' for label in value_counts.index]

            patches, texts = ax_pie.pie(
                value_counts,
                startangle=90,
                colors=pie_colors_sns,
                wedgeprops={'edgecolor': 'white'}
            )
            ax_pie.set_title(f'Distribuição Percentual de {col} (Pizza)', fontsize=16)
            ax_pie.axis('equal')

            num_legend_cols = 1
            if value_counts.shape[0] > 2: num_legend_cols = 2
            if value_counts.shape[0] > 4: num_legend_cols = 3
            if value_counts.shape[0] > 6: num_legend_cols = 4


            ax_pie.legend(patches, labels_for_legend,
                          title=f"Categorias de {col}",
                          loc='upper center',
                          bbox_to_anchor=(0.5, 0.05), # Ajustado para posicionar abaixo
                          fontsize=9,
                          ncol=num_legend_cols,
                          frameon=True,
                          shadow=True)

            # Ajustar o 'bottom' para dar espaço para a legenda
            # O valor exato pode depender do número de itens na legenda e do tamanho da fonte
            if num_legend_cols == 1:
                plt.subplots_adjust(bottom=0.1)
            elif num_legend_cols == 2:
                plt.subplots_adjust(bottom=0.15)
            else: # 3 ou 4 colunas
                plt.subplots_adjust(bottom=0.2)


            plt.savefig(os.path.join(eda_output_dir_script, f'univar_cat_{col}_piechart_bottom_legend.png'))
            plt.show()
            plt.close(fig_pie)
            print(f"  Gráfico de pizza com legenda inferior para '{col}' salvo.")

        if value_counts.shape[0] > 0 :
            try:
                df_plotly_bar = value_counts.reset_index()
                df_plotly_bar.columns = ['Categoria', 'Contagem']
                current_palette_plotly = cat_palettes_plotly[i % len(cat_palettes_plotly)]
                fig_plotly = px.bar(df_plotly_bar,
                                    y='Categoria', x='Contagem',
                                    orientation='h', title=f'Distribuição Interativa de {col}',
                                    text='Contagem', color='Categoria',
                                    color_discrete_sequence=current_palette_plotly)
                fig_plotly.update_layout(yaxis_title=col, xaxis_title='Contagem', yaxis={'categoryorder':'total ascending'})
                if value_counts.shape[0] == 1:
                    fig_plotly.update_traces(marker_color='lightgreen')
                fig_plotly.update_traces(texttemplate='%{text}', textposition='outside')
                fig_plotly.write_html(os.path.join(eda_output_dir_script, f'univar_cat_{col}_plotly_bar.html'))
                fig_plotly.show()
                print(f"  Gráfico interativo de barras para '{col}' salvo como HTML e exibido.")
            except Exception as e:
                print(f"  Não foi possível gerar gráfico Plotly para {col}: {e}")
    else:
        print(f"  Coluna '{col}' não encontrada ou sem categorias únicas. Pulando análise categórica.")

print(f"\n--- Análise Univariada Concluída. Gráficos salvos em: {os.path.abspath(eda_output_dir_script)} ---")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re
import plotly.express as px
import statsmodels.api as sm
from statsmodels.graphics.mosaicplot import mosaic




# --- 7. Análise Bivariada (COM NOVAS CORES) ---
print("\n--- 7. Análise Bivariada ---")
target_eda_col = 'faixa_salarial_eda_2cat'
numeric_cols_for_bivar_analysis = ['salary_numeric_lower_bound', 'experiencia_anos'] # Renomeado para evitar conflito
categorical_features_for_bivar_analysis = ['P1_a_1', 'P1_b', 'P1_l', 'P2_f', 'P2_g', 'Regiao_Mapeada']

# Definindo paletas para a análise bivariada
palette_target_binary = sns.color_palette(["#3498db", "#f1c40f"]) # Azul, Amarelo para alvo binário
palette_cat_vs_cat = "viridis_r" # Exemplo de paleta para categórica vs categórica
palette_num_vs_num_hue = "coolwarm" # Para scatterplots com hue

if target_eda_col in df_eda.columns and df_eda[target_eda_col].nunique() > 0:

    # 7.1 Feature Numérica vs. Variável Alvo Categórica
    print("\n  7.1 Feature Numérica vs. Variável Alvo Categórica")
    for num_col in numeric_cols_for_bivar_analysis:
        if num_col in df_eda.columns and pd.api.types.is_numeric_dtype(df_eda[num_col]):
            print(f"\n    Analisando '{num_col}' vs '{target_eda_col}':")

            plt.figure(figsize=(10, 6))
            sns.boxplot(x=target_eda_col, y=num_col, data=df_eda, palette=palette_target_binary)
            plt.title(f'Boxplot de {num_col} por {target_eda_col}', fontsize=15); plt.tight_layout()
            plt.savefig(os.path.join(eda_output_dir_script, f'bivar_boxplot_{num_col}_vs_{target_eda_col}.png')); plt.show(); plt.close()

            plt.figure(figsize=(10, 6))
            sns.violinplot(x=target_eda_col, y=num_col, data=df_eda, palette=palette_target_binary)
            plt.title(f'Violin Plot de {num_col} por {target_eda_col}', fontsize=15); plt.tight_layout()
            plt.savefig(os.path.join(eda_output_dir_script, f'bivar_violin_{num_col}_vs_{target_eda_col}.png')); plt.show(); plt.close()

            plt.figure(figsize=(10, 6))
            sns.histplot(data=df_eda, x=num_col, hue=target_eda_col, kde=True, multiple="layer", palette=palette_target_binary, bins=30, alpha=0.6)
            plt.title(f'Histograma Sobreposto de {num_col} por {target_eda_col}', fontsize=15); plt.tight_layout()
            plt.savefig(os.path.join(eda_output_dir_script, f'bivar_hist_layer_{num_col}_vs_{target_eda_col}.png')); plt.show(); plt.close()
            print(f"    Gráficos para '{num_col}' vs '{target_eda_col}' salvos.")

    # 7.2 Feature Categórica vs. Variável Alvo Categórica
    print("\n  7.2 Feature Categórica vs. Variável Alvo Categórica")
    for cat_col in categorical_features_for_bivar_analysis:
        if cat_col in df_eda.columns:
            print(f"\n    Analisando '{cat_col}' vs '{target_eda_col}':")
            order = df_eda[cat_col].value_counts().nlargest(15).index

            plt.figure(figsize=(12, max(7, len(order) * 0.5)))
            sns.countplot(y=cat_col, hue=target_eda_col, data=df_eda[df_eda[cat_col].isin(order)], order=order, palette=palette_target_binary)
            plt.title(f'Contagem de {target_eda_col} por {cat_col} (Top {len(order)})', fontsize=15); plt.legend(title=target_eda_col); plt.tight_layout()
            plt.savefig(os.path.join(eda_output_dir_script, f'bivar_countplot_{cat_col}_vs_{target_eda_col}.png')); plt.show(); plt.close()

            props = df_eda.groupby(cat_col)[target_eda_col].value_counts(normalize=True).mul(100).rename('percentage').reset_index()
            props_pivot = props.pivot(index=cat_col, columns=target_eda_col, values='percentage').fillna(0)
            if not props_pivot.empty:
                categories_in_both = [c for c in order if c in props_pivot.index]
                props_pivot_filtered = props_pivot.loc[categories_in_both]
                if not props_pivot_filtered.empty:
                    props_pivot_filtered.plot(kind='barh', stacked=True,
                                              figsize=(12, max(7, len(props_pivot_filtered) * 0.5)),
                                              color=[palette_target_binary[0], palette_target_binary[1]] if len(props_pivot_filtered.columns) == 2 else palette_target_binary[0]) # Cores para empilhado
                    plt.title(f'Proporção de {target_eda_col} por {cat_col} (Top {len(order)})', fontsize=15)
                    plt.xlabel('Porcentagem (%)', fontsize=12); plt.ylabel(cat_col, fontsize=12)
                    plt.legend(title=target_eda_col, bbox_to_anchor=(1.05, 1), loc='upper left'); plt.tight_layout(rect=[0,0,0.85,1])
                    plt.savefig(os.path.join(eda_output_dir_script, f'bivar_stacked_prop_{cat_col}_vs_{target_eda_col}.png')); plt.show(); plt.close()

            contingency_table = pd.crosstab(df_eda[cat_col], df_eda[target_eda_col])
            plt.figure(figsize=(10, max(6, len(contingency_table) * 0.4)))
            sns.heatmap(contingency_table, annot=True, fmt="d", cmap="BuPu") # Paleta BuPu para heatmap
            plt.title(f'Tabela de Contingência: {cat_col} vs {target_eda_col}', fontsize=15); plt.tight_layout()
            plt.savefig(os.path.join(eda_output_dir_script, f'bivar_heatmap_{cat_col}_vs_{target_eda_col}.png')); plt.show(); plt.close()
            print(f"    Gráficos para '{cat_col}' vs '{target_eda_col}' salvos.")

    # 7.3 Feature Numérica vs. Feature Numérica
    print("\n  7.3 Feature Numérica vs. Feature Numérica")
    if 'experiencia_anos' in df_eda.columns and 'salary_numeric_lower_bound' in df_eda.columns:
        plt.figure(figsize=(10, 7))
        sns.scatterplot(x='experiencia_anos', y='salary_numeric_lower_bound',
                        hue=target_eda_col if target_eda_col in df_eda.columns else None,
                        data=df_eda, alpha=0.7, palette=palette_target_binary) # Usando paleta do alvo
        plt.title('Experiência vs. Salário por Faixa Salarial', fontsize=15)
        plt.xlabel('Experiência (anos)', fontsize=12); plt.ylabel('Limite Inferior do Salário', fontsize=12)
        plt.legend(title=target_eda_col if target_eda_col in df_eda.columns else None); plt.tight_layout()
        plt.savefig(os.path.join(eda_output_dir_script, 'bivar_scatter_exp_vs_salario.png')); plt.show(); plt.close()

        if not df_eda[['experiencia_anos', 'salary_numeric_lower_bound']].dropna().empty:
            g = sns.jointplot(x='experiencia_anos', y='salary_numeric_lower_bound', data=df_eda,
                              kind="kde", fill=True, cmap="YlGnBu", # Paleta YlGnBu para KDE
                              hue=target_eda_col if target_eda_col in df_eda.columns else None,
                              palette=palette_target_binary if target_eda_col in df_eda.columns else None)
            g.fig.suptitle('Joint Plot: Experiência vs Salário (KDE)', y=1.02, fontsize=15); g.fig.tight_layout(rect=[0,0,1,0.98])
            plt.savefig(os.path.join(eda_output_dir_script, 'bivar_jointplot_kde_exp_vs_salario.png')); plt.show(); plt.close()

        cols_for_pairplot = [col for col in numeric_cols_for_bivar_analysis if col in df_eda.columns]
        if len(cols_for_pairplot) >=2:
            pair_plot_df = df_eda[cols_for_pairplot + ([target_eda_col] if target_eda_col in df_eda.columns else [])].dropna()
            if not pair_plot_df.empty and (not target_eda_col in df_eda.columns or pair_plot_df[target_eda_col].nunique() > 0):
                sns.pairplot(pair_plot_df, hue=target_eda_col if target_eda_col in df_eda.columns else None,
                             palette=palette_target_binary if target_eda_col in df_eda.columns else None, diag_kind="kde", plot_kws={'alpha': 0.6})
                plt.suptitle('Pair Plot de Variáveis Numéricas', y=1.02, fontsize=15); plt.tight_layout(rect=[0,0,1,0.98])
                plt.savefig(os.path.join(eda_output_dir_script, 'bivar_pairplot_numeric.png')); plt.show(); plt.close()

    # 7.4 Feature Categórica vs. Feature Categórica (Não Alvo)
    print("\n  7.4 Feature Categórica vs. Feature Categórica (Não Alvo)")
    if 'P1_l' in df_eda.columns and 'P2_g' in df_eda.columns:
        cat1, cat2 = 'P1_l', 'P2_g'
        print(f"\n    Analisando '{cat1}' vs '{cat2}':")
        order1 = df_eda[cat1].value_counts().nlargest(10).index; order2 = df_eda[cat2].value_counts().nlargest(5).index
        df_subset = df_eda[(df_eda[cat1].isin(order1)) & (df_eda[cat2].isin(order2))]
        if not df_subset.empty:
            plt.figure(figsize=(12, 8))
            palette_cat_cat_bivar = sns.color_palette("tab20", n_colors=df_subset[cat2].nunique()) # Paleta tab20
            sns.countplot(data=df_subset, y=cat1, hue=cat2, order=order1, hue_order=order2, palette=palette_cat_cat_bivar)
            plt.title(f'Distribuição de {cat2} por {cat1} (Top Categorias)', fontsize=15)
            plt.legend(title=cat2, bbox_to_anchor=(1.05, 1), loc='upper left'); plt.tight_layout(rect=[0,0,0.85,1])
            plt.savefig(os.path.join(eda_output_dir_script, f'bivar_countplot_{cat1}_vs_{cat2}.png')); plt.show(); plt.close()

            cross_tab_mosaic = pd.crosstab(df_subset[cat1], df_subset[cat2])
            if not cross_tab_mosaic.empty and cross_tab_mosaic.sum().sum() > 0:
                try:
                    props = {} # Para customizar cores no mosaic plot, pode ser complexo
                    # Exemplo: props = lambda key: {'color': 'skyblue' if 'Júnior' in key else 'lightcoral'}
                    fig_mosaic, _ = mosaic(df_subset, [cat1, cat2], title=f'Mosaic Plot: {cat1} vs {cat2}', gap=0.02, properties=props)
                    fig_mosaic.set_size_inches(12, 8); plt.tight_layout()
                    plt.savefig(os.path.join(eda_output_dir_script, f'bivar_mosaic_{cat1}_vs_{cat2}.png')); plt.show(); plt.close()
                except Exception as e: print(f"    Erro ao gerar mosaic plot para {cat1} vs {cat2}: {e}")
        else: print(f"    Não há dados suficientes após filtrar top categorias para '{cat1}' vs '{cat2}'.")
else:
    print(f"AVISO: Coluna alvo '{target_eda_col}' não encontrada ou sem categorias válidas, algumas análises bivariadas serão puladas.")

print(f"\n--- Análise Bivariada Concluída (Exemplos). Gráficos salvos em: {os.path.abspath(eda_output_dir_script)} ---")
```

```python
# (ASSUMA QUE AS SEÇÕES 1-7 DA EDA, INCLUINDO CARREGAMENTO, PRÉ-PROCESSAMENTO,
# ANÁLISE UNIVARIADA E BIVARIADA JÁ FORAM EXECUTADAS)
# df_eda está pronto e contém as colunas processadas.
# eda_output_dir_script também já está definido.

print("\n--- 8. Análise Multivariada ---")

# Reutilizar ou definir variáveis e paletas se não estiverem no escopo global
# ou se quiser novas para esta seção.
# Para este exemplo, vamos assumir que já existem do bloco bivariado:
target_eda_col = 'faixa_salarial_eda_2cat'
numeric_cols_for_bivar_analysis = ['salary_numeric_lower_bound', 'experiencia_anos']
categorical_features_for_bivar_analysis = ['P1_a_1', 'P1_b', 'P1_l', 'P2_f', 'P2_g', 'Regiao_Mapeada']
bivar_palette_target_distinct = "Set1"
bivar_palette_scatter_hue_bivar = "plasma"

# Novas paletas específicas para multivariada, se necessário:
multi_palette_hue_cat1 = "viridis"  # Exemplo para uma categórica no hue
multi_palette_hue_cat2 = "coolwarm" # Exemplo para outra categórica no hue

# 8.1 Relação entre Duas Variáveis Numéricas, Segmentada por uma Categórica
print("\n  8.1 Duas Numéricas vs. Uma Categórica")

if 'experiencia_anos' in df_eda.columns and 'salary_numeric_lower_bound' in df_eda.columns:
    # Experiência vs Salário, segmentado por Nível de Senioridade (P2_g)
    if 'P2_g' in df_eda.columns:
        plt.figure(figsize=(14, 8))
        top_senioridade_multi = df_eda['P2_g'].value_counts().nlargest(5).index
        df_subset_senioridade_multi = df_eda[df_eda['P2_g'].isin(top_senioridade_multi)]

        if not df_subset_senioridade_multi.empty:
            sns.scatterplot(data=df_subset_senioridade_multi,
                            x='experiencia_anos',
                            y='salary_numeric_lower_bound',
                            hue='P2_g',
                            size='P2_g',
                            palette=multi_palette_hue_cat1,
                            alpha=0.7,
                            hue_order=top_senioridade_multi,
                            size_order=top_senioridade_multi,
                            sizes=(40, 200))
            plt.title('Experiência vs. Salário por Nível de Senioridade', fontsize=16)
            plt.xlabel('Experiência (anos)', fontsize=12)
            plt.ylabel('Limite Inferior do Salário', fontsize=12)
            plt.legend(title='Nível de Senioridade', bbox_to_anchor=(1.05, 1), loc='upper left')
            plt.tight_layout(rect=[0,0,0.85,1]) # Ajusta para caber legenda
            plt.savefig(os.path.join(eda_output_dir_script, 'multivar_scatter_exp_sal_senioridade.png'))
            plt.show()
            plt.close()
            print("    Gráfico Scatter: Experiência vs Salário por Senioridade salvo.")

    # Experiência vs Salário, segmentado por Nível de Ensino (P1_l) e Faixa Salarial (Alvo)
    if 'P1_l' in df_eda.columns and target_eda_col in df_eda.columns:
        top_ensino_multi = df_eda['P1_l'].value_counts().nlargest(4).index
        df_subset_ensino_multi = df_eda[df_eda['P1_l'].isin(top_ensino_multi)]

        if not df_subset_ensino_multi.empty and df_subset_ensino_multi[target_eda_col].nunique() > 0:
            # Garantir que a paleta bivar_palette_target_distinct está definida
            if 'bivar_palette_target_distinct' not in locals():
                 bivar_palette_target_distinct = "Set1" # Fallback

            plt.figure(figsize=(14, 8))
            sns.scatterplot(data=df_subset_ensino_multi,
                            x='experiencia_anos',
                            y='salary_numeric_lower_bound',
                            hue=target_eda_col,
                            style='P1_l',
                            palette=bivar_palette_target_distinct,
                            hue_order=sorted(df_eda[target_eda_col].unique()),
                            style_order=top_ensino_multi,
                            s=100,
                            alpha=0.8)
            plt.title('Experiência vs. Salário por Nível de Ensino e Faixa Salarial', fontsize=16)
            plt.xlabel('Experiência (anos)', fontsize=12)
            plt.ylabel('Limite Inferior do Salário', fontsize=12)
            plt.legend(title='Legenda', bbox_to_anchor=(1.05, 1), loc='upper left')
            plt.tight_layout(rect=[0,0,0.85,1])
            plt.savefig(os.path.join(eda_output_dir_script, 'multivar_scatter_exp_sal_ensino_alvo.png'))
            plt.show()
            plt.close()
            print("    Gráfico Scatter: Experiência vs Salário por Ensino e Faixa Salarial salvo.")

# 8.2 Distribuição de uma Variável Numérica, Segmentada por Duas Categóricas
print("\n  8.2 Uma Numérica vs. Duas Categóricas")
if 'salary_numeric_lower_bound' in df_eda.columns and target_eda_col in df_eda.columns:
    # Salário por Nível de Ensino (P1_l), segmentado pela Faixa Salarial (Alvo)
    if 'P1_l' in df_eda.columns:
        plt.figure(figsize=(14, 8))
        order_ensino = df_eda['P1_l'].value_counts().nlargest(7).index
        df_subset_ensino_target = df_eda[df_eda['P1_l'].isin(order_ensino)]

        if not df_subset_ensino_target.empty and df_subset_ensino_target[target_eda_col].nunique() > 0:
            if 'bivar_palette_target_distinct' not in locals(): bivar_palette_target_distinct = "Set1"

            sns.boxplot(data=df_subset_ensino_target,
                        x='P1_l',
                        y='salary_numeric_lower_bound',
                        hue=target_eda_col,
                        order=order_ensino,
                        palette=bivar_palette_target_distinct)
            plt.title('Salário por Nível de Ensino e Faixa Salarial (Alvo)', fontsize=16)
            plt.xlabel('Nível de Ensino (P1_l)', fontsize=12)
            plt.ylabel('Limite Inferior do Salário', fontsize=12)
            plt.xticks(rotation=45, ha='right')
            plt.legend(title='Faixa Salarial (Alvo)')
            plt.tight_layout()
            plt.savefig(os.path.join(eda_output_dir_script, 'multivar_boxplot_sal_ensino_alvo.png'))
            plt.show()
            plt.close()
            print("    Gráfico Boxplot: Salário por Nível de Ensino e Faixa Salarial (Alvo) salvo.")

    # Experiência por Senioridade, segmentado pela Faixa Salarial (Alvo)
    if 'P2_g' in df_eda.columns and 'experiencia_anos' in df_eda.columns:
        plt.figure(figsize=(12, 7))
        order_senioridade = df_eda['P2_g'].value_counts().nlargest(5).index
        df_subset_senior_target = df_eda[df_eda['P2_g'].isin(order_senioridade)]

        if not df_subset_senior_target.empty and df_subset_senior_target[target_eda_col].nunique() > 0:
            if 'bivar_palette_target_distinct' not in locals(): bivar_palette_target_distinct = "Set1"

            sns.violinplot(data=df_subset_senior_target,
                           x='P2_g',
                           y='experiencia_anos',
                           hue=target_eda_col,
                           order=order_senioridade,
                           palette=bivar_palette_target_distinct,
                           split=True)
            plt.title('Experiência por Senioridade e Faixa Salarial (Alvo)', fontsize=16)
            plt.xlabel('Nível de Senioridade (P2_g)', fontsize=12)
            plt.ylabel('Experiência (anos)', fontsize=12)
            plt.xticks(rotation=45, ha='right')
            plt.legend(title='Faixa Salarial (Alvo)')
            plt.tight_layout()
            plt.savefig(os.path.join(eda_output_dir_script, 'multivar_violin_exp_senior_alvo.png'))
            plt.show()
            plt.close()
            print("    Gráfico Violin: Experiência por Senioridade e Faixa Salarial (Alvo) salvo.")

# 8.3 Relação entre Três Variáveis Categóricas
print("\n  8.3 Três Variáveis Categóricas")
if 'Regiao_Mapeada' in df_eda.columns and 'P1_l' in df_eda.columns and target_eda_col in df_eda.columns:
    top_regioes_facet = df_eda['Regiao_Mapeada'].value_counts().nlargest(4).index
    top_ensino_facet = df_eda['P1_l'].value_counts().nlargest(5).index
    df_subset_facet = df_eda[
        df_eda['Regiao_Mapeada'].isin(top_regioes_facet) &
        df_eda['P1_l'].isin(top_ensino_facet)
    ]

    if not df_subset_facet.empty and df_subset_facet['P1_l'].nunique() > 0 and df_subset_facet[target_eda_col].nunique() > 0:
        if 'bivar_palette_target_distinct' not in locals(): bivar_palette_target_distinct = "Set1"

        g = sns.catplot(data=df_subset_facet,
                        y='P1_l',
                        hue=target_eda_col,
                        col='Regiao_Mapeada',
                        kind='count',
                        palette=bivar_palette_target_distinct,
                        height=6, aspect=0.9, # Ajusta tamanho e proporção das facetas
                        order=top_ensino_facet,
                        col_order = top_regioes_facet,
                        sharey=False)
        g.fig.suptitle('Nível de Ensino por Região e Faixa Salarial (Alvo)', y=1.03, fontsize=16)
        g.set_axis_labels("Contagem", "Nível de Ensino (P1_l)")
        g.set_titles("Região: {col_name}")
        g.set_xticklabels(rotation=45) # Rotacionar labels do eixo x se necessário nas facetas
        plt.tight_layout(rect=[0,0,1,0.97])
        plt.savefig(os.path.join(eda_output_dir_script, 'multivar_catplot_ensino_regiao_alvo.png'))
        plt.show()
        plt.close()
        print("    Gráfico Catplot (FacetGrid): Nível de Ensino por Região e Faixa Salarial (Alvo) salvo.")

    # Plotly para interatividade (exemplo: Faixa Etária, Região e Faixa Salarial)
    if 'P1_a_1' in df_eda.columns:
        top_idade_plotly = df_eda['P1_a_1'].value_counts().nlargest(5).index
        df_subset_plotly_cat = df_eda[
            df_eda['Regiao_Mapeada'].isin(top_regioes_facet) &
            df_eda['P1_a_1'].isin(top_idade_plotly)
        ]
        if not df_subset_plotly_cat.empty:
            try:
                df_agg_plotly = df_subset_plotly_cat.groupby(['Regiao_Mapeada', 'P1_a_1', target_eda_col]).size().reset_index(name='Contagem')
                if not df_agg_plotly.empty:
                    fig_plotly_multi = px.bar(df_agg_plotly,
                                            x='P1_a_1',
                                            y='Contagem',
                                            color=target_eda_col,
                                            facet_col='Regiao_Mapeada',
                                            title='Contagem por Faixa Etária, Região e Faixa Salarial',
                                            labels={'P1_a_1': 'Faixa Etária', target_eda_col: 'Faixa Salarial'},
                                            category_orders={"Regiao_Mapeada": top_regioes_facet.tolist(),
                                                             "P1_a_1": top_idade_plotly.tolist()},
                                            color_discrete_map={"Salário Baixo": "#636EFA", "Salário Alto": "#FFA15A"} # Paleta Plotly
                                            )
                    fig_plotly_multi.update_xaxes(categoryorder='array', categoryarray=top_idade_plotly.tolist())
                    fig_plotly_multi.update_layout(height=600)
                    html_path = os.path.join(eda_output_dir_script, 'multivar_plotly_idade_regiao_alvo.html')
                    png_path = os.path.join(eda_output_dir_script, 'multivar_plotly_idade_regiao_alvo.png')
                    fig_plotly_multi.write_html(html_path)
                    try: pio.write_image(fig_plotly_multi, png_path, width=1200, height=700)
                    except Exception as e_img: print(f"    Falha ao salvar Plotly como PNG: {e_img} (necessita 'kaleido')")
                    fig_plotly_multi.show()
                    print("    Gráfico Plotly multivariado (Idade, Região, Alvo) salvo e exibido.")
            except Exception as e:
                print(f"    Erro ao gerar gráfico Plotly multivariado (Idade, Região, Alvo): {e}")

# 8.4 Heatmap de Correlação (Incluindo Alvo Numérico)
print("\n  8.4 Heatmap de Correlação com Alvo Numérico")
numeric_cols_for_corr_multi = [col for col in ['salary_numeric_lower_bound', 'experiencia_anos'] if col in df_eda.columns]
target_for_corr_multi = target_eda_col # Usar o nome definido para a coluna alvo

if numeric_cols_for_corr_multi and target_for_corr_multi in df_eda.columns:
    df_eda_corr_multi = df_eda[numeric_cols_for_corr_multi].copy()
    le_eda_multi = LabelEncoder()
    df_eda_corr_multi['alvo_numerico_eda'] = le_eda_multi.fit_transform(df_eda[target_for_corr_multi])
    print(f"    Classes do alvo numérico '{target_for_corr_multi}' para correlação: {list(le_eda_multi.classes_)} -> {list(le_eda_multi.transform(le_eda_multi.classes_))}")

    cols_for_heatmap_multi = numeric_cols_for_corr_multi + ['alvo_numerico_eda']
    if all(col in df_eda_corr_multi.columns for col in cols_for_heatmap_multi): # Checa se 'alvo_numerico_eda' foi adicionado
        corr_matrix_multi = df_eda_corr_multi[cols_for_heatmap_multi].corr()
        plt.figure(figsize=(8, 6))
        sns.heatmap(corr_matrix_multi, annot=True, cmap="RdYlBu", fmt=".2f", linewidths=.5, vmin=-1, vmax=1) # Paleta RdYlBu
        plt.title('Heatmap de Correlação (Incluindo Alvo Numérico)', fontsize=15)
        plt.tight_layout()
        plt.savefig(os.path.join(eda_output_dir_script, 'multivar_corr_heatmap_with_target.png'))
        plt.show()
        plt.close()
        print("    Heatmap de Correlação (com alvo) salvo.")
    else:
        print("    Nem todas as colunas necessárias para o heatmap de correlação com alvo estão disponíveis em df_eda_corr_multi.")
else:
    print("    Colunas numéricas ou coluna alvo não encontradas para o heatmap de correlação.")

print(f"\n--- Análise Multivariada Concluída (Exemplos). Gráficos salvos em: {os.path.abspath(eda_output_dir_script)} ---")
```
